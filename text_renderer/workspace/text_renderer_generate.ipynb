```python
vocab = set()
with open('corpus/google_sp.vocab', 'r') as f:
    for line in f.readlines():
        vocab.add(line[0])
        
vocab = list(vocab)
print(vocab[100:120])
print(len(vocab))
```

```python
with open('corpus/charset.txt', 'r') as fp:
    text = fp.read()

text=text.replace('\n', '')

with open('corpus/charset.txt', 'w') as fp:
    fp.write(text)
    
```

```python
newtext = ''
for char in vocab:
    newtext += char
    newtext += '\n'
print(len(newtext))
with open('corpus/charset.txt', 'w') as fp:
    fp.write(newtext)
```

```python
import lzma
text = str()
with lzma.open('corpus/ja.txt.xz', mode='rt', encoding='utf-8') as fp:
    for _ in range(2*(10**7)):
        text += fp.readline()
    
    
print(len(text))

with open('corpus/corpus.txt', 'w') as fp:
    fp.write(text)
```
